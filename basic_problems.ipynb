```python
# Build a simple calculator for addition, subtraction, multiplication, division
def get_calculation():
    """ Take user input and validate it
    """
    
    calculation = input("Enter calculation with format { num op num } (eg. 1 + 1):\n").split()
    if len(calculation) != 3:
        print("\nInvalid calculation! Correct format is { num op num }")
        return None
    if calculation[1] not in ('+', '-', '*', '/'):
        print(f"\nInvalid operator '{calculation[1]}'!")
        return None

    num1 = int(calculation[0])
    op = calculation[1]
    num2 = int(calculation[2])

    return num1, op, num2

def calculate(calculation):
    num1, op, num2 = calculation

    match op:
        case '+':
            result = num1 + num2
        case '-':
            result = num1 - num2
        case '*':
            result = num1 * num2
        case '/':
            if num2 == 0:
                print("Cannot divide by zero!")
                return None
            else:
                result = num1 / num2
        case _:
            return None

    return result

calculation = get_calculation()
if calculation is not None:
    result = calculate(calculation)
    if result is not None:
        print(f"\nResult = {result}")
```

```python
# check if given number is positive, negative, or zero

num = int(input("Enter a number: "))
sign_str = "Given number is %s."

if num == 0:
    print(sign_str % "zero")
elif num > 0:
    print(sign_str % "positive")
else:
    print(sign_str % "negative")
```

```python
# print sum of all digits of number

num_str = input("Enter a number: ")
num_int = int(num_str)
sum = 0

print(f"{num_str} has {len(num_str)} digits.")
for i in range(1, len(num_str) + 1):
    sum += (num_int % (10**i)) // (10**(i-1))

print(f"\nSum of digits = {sum}")
```

```python
# find factorial of a given number

num = int(input("Enter a number: "))

def factorial(num):
    if num == 0:
        return 1
    elif num <= 0:
        print("Cannot calculate factorial of a negative number!")
        return None

    factorial = 1
    for i in range(1, num + 1):
        factorial *= i

    return factorial

fac = factorial(num)
if fac is not None:
    print(f"{num}! = {fac}")
```

```python
# check if number is prime or not
num = int(input("Enter a number: "))

def check_prime(num):
    if num <= 1:
        print(f"{num} is neither prime nor composite")
        return None

    for i in range(2, num // 2):
        if num % i == 0:
            return False

    return True

is_prime = check_prime(num)

if is_prime is not None:
    if is_prime:
        print(f"{num} is a prime number.")
    else:
        print(f"{num} is not a prime number.")
```

```python
# print prime numbers between 10 and 99
print("Prime number between 10 and 99 are as follows:")

for i in range(10, 100):
    if check_prime(i):
        print(i, end=" ")
```
